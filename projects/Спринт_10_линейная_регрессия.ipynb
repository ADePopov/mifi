### Напишите функцию для стандартизации признаков. Расчёт параметров масштабирования делайте только по обучающей выборке, чтобы не дать утечке ни малейшего шанса.
### Напишите функцию для нормализации признаков.

```python
def scale_data(X_train, X_val, X_test, method='standard'):
    if method == 'standard':
        mean = X_train.mean()
        std = X_train.std()
        X_train_scaled = (X_train - mean) / std
        X_val_scaled = (X_val - mean) / std
        X_test_scaled = (X_test - mean) / std
    elif method == 'minmax':
        min_val = X_train.min()
        max_val = X_train.max()
        X_train_scaled = (X_train - min_val) / (max_val - min_val)
        X_val_scaled = (X_val - min_val) / (max_val - min_val)
        X_test_scaled = (X_test - min_val) / (max_val - min_val)
    else:
        raise ValueError("Неверный метод масштабирования. Используйте 'standard' или 'minmax'.")
    return X_train_scaled, X_val_scaled, X_test_scaled
```

**Вывод**

Разработана универсальная функция scale_data, которая поддерживает два метода масштабирования:

Стандартизация (Z-score): (X - mean) / std — приводит признаки к распределению со средним 0 и стандартным отклонением 1.

Нормализация (MinMax): (X - min) / (max - min) — масштабирует признаки в диапазон.

Ключевое требование выполнено: расчёт параметров масштабирования (среднее, стандартное отклонение, минимум, максимум) производится строго по обучающей выборке (X_train). Эти же параметры затем применяются к валидационной и тестовой выборкам.

Такой подход полностью исключает утечку данных (data leakage) — информация из валидационной и тестовой выборок не участвует в вычислении параметров масштабирования, что гарантирует честную оценку качества модели на новых, невидимых данных.

### Обучите несколько архитектур линейных моделей. Они могут различаться по ряду черт: набором отобранных признаков, масштабом признаков, установленными гиперпараметрами, функциями потерь. Попробуйте обучить следующие модели:
   - `LinearRegression`;
   - `Lasso` (L1-регуляризация);
   - `Ridge` (L2-регуляризация);
   - `SGDRegressor`.
   - **Бонусное задание.** Подумайте, можно ли улучшить модели за счёт создания новых признаков: например, умножив длину ласт на ширину. Проверьте, усилится ли корреляция нового признака с целевой переменной, возрастёт ли благодаря ему качество модели.

```python
def calculate_metrics(y_true, y_pred):
    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mse = mean_squared_error(y_true, y_pred)
    mae = mean_absolute_error(y_true, y_pred)
    mape = mean_absolute_percentage_error(y_true, y_pred)
    r2 = r2_score(y_true, y_pred)
    return {'RMSE': round(rmse, 2),
        'MSE': round(mse, 2),
        'MAPE': round(mape * 100, 2),
        'MAE': round(mae, 2),
        'R2': round(r2, 3)}

def compare_metrics(train_metrics, val_metrics):
    for metric in train_metrics:
        change = (val_metrics[metric] - train_metrics[metric]) / train_metrics[metric] * 100
        direction = "изменилась" if metric != 'R2' else "изменился"
        print(f"{metric} на val {direction} по сравнению с {metric} на train на {change:.2f}%")
```

```python
def evaluate_model(model, X_train, X_val, X_test, y_train, y_val, y_test, model_name=None):
    model.fit(X_train, y_train)
    y_train_pred = model.predict(X_train)
    y_val_pred = model.predict(X_val)
    y_test_pred = model.predict(X_test)
    metrics = {
        'model': model, 
        'model_name': model_name,
        'train': calculate_metrics(y_train, y_train_pred),
        'val': calculate_metrics(y_val, y_val_pred),
        'test': calculate_metrics(y_test, y_test_pred)
    }
    return metrics
def print_metrics(metrics):
    print(f"Модель: {metrics['model_name']}")
    print("Обучающая выборка:")
    for key, value in metrics['train'].items():
        print(f"  {key}: {value}")
    print("Валидационная выборка:")
    for key, value in metrics['val'].items():
        print(f"  {key}: {value}")
    print("Тестовая выборка:")
    for key, value in metrics['test'].items():
        print(f"  {key}: {value}")
```

#### LinearRegression

```python
lr_raw = evaluate_model(LinearRegression(), X_train_raw, X_val_raw, X_test_raw,
             y_train, y_val, y_test,model_name = 'LinearRegression_raw')
lr_norm = evaluate_model(LinearRegression(), X_train_norm, X_val_norm, X_test_norm,
             y_train, y_val, y_test,model_name = 'LinearRegression_norm')
lr_std = evaluate_model(LinearRegression(), X_train_std, X_val_std, X_test_std,
             y_train, y_val, y_test,model_name = 'LinearRegression_std')
print_metrics(lr_raw)
print_metrics(lr_norm)
print_metrics(lr_std)
```

```python
weights = best_model.coef_
intercept = best_model.intercept_
feature_names = X_train_raw.columns

def predict_weight_turtles(features, model):
    if isinstance(features, dict):
        X = np.array([features[name] for name in feature_names]).reshape(1, -1)
    else:
        X = np.array(features).reshape(1, -1)
    return model.predict(X)[0]
```
