```python
##### eu sales
mean_eu_sales_platform = df.groupby(['year_of_release','platform'])['eu_sales'].mean()


def full_sales(row):
    if pd.isna(row['eu_sales']):
        return mean_eu_sales_platform[(row['year_of_release'],row['platform'])]
    else:
        return row['eu_sales']
df['eu_sales'] = df.apply(full_sales, axis = 1)
```

```python
##### jp sales
mean_jp_sales_platform = df.groupby(['year_of_release','platform'])['jp_sales'].mean()

def full_sales(row):
    if pd.isna(row['jp_sales']):
        return mean_jp_sales_platform[(row['year_of_release'],row['platform'])]
    else:
        return row['jp_sales']
df['jp_sales'] = df.apply(full_sales, axis = 1)
```
